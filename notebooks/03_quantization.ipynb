```python
import subprocess
import sys
import time

import requests

server = subprocess.Popen(
    [sys.executable, "-m", "uvicorn", "vlm_lab.api:app", "--host", "127.0.0.1", "--port", "8000"],
    cwd="..",
)
for _ in range(30):
    try:
        if requests.get("http://127.0.0.1:8000/api/v1/health").ok:
            break
    except requests.RequestException:
        time.sleep(1)
```

```python
import json

with open("../configs/benchmark_quick.json", encoding="utf-8") as handle:
    config = json.load(handle)
config["dataset_manifest"] = str(
    __import__("pathlib").Path("../data/processed/flickr8k_test.csv").resolve()
)
run = requests.post("http://127.0.0.1:8000/api/v1/experiments", json=config).json()
run
```
