```python
import json

import pandas as pd
from PIL import Image

from vlm_lab.domain import GenerationOptions, Precision
from vlm_lab.inference import CaptionEngine

row = pd.read_csv("../data/processed/flickr8k_test.csv").iloc[0]
image = Image.open(row.image_path).convert("RGB")
display(image)
print("Références:", json.loads(row.references))
```

```python
engine = CaptionEngine(cache_size=1)
for model_id in ["blip-base", "git-base"]:
    result = engine.caption(image, model_id, Precision.FULL, GenerationOptions())
    print(model_id, result.model_dump())
```
